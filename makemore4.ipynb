```python
#  with 27 character, if everything was equally likely, then you expect all of this probability to be 
1/27 
# 4% roughly: anything above 4% means we have learns something useful
# 4*27
# how can we summarize the probabilities above into a single number that measures the quakity of this model? [maximum likelihood e stimation and statisitcal modelling]:

# the product of the probability is the likelihood, with a good model, the product should be too large

# beacuase our product can vanish to 0, we use the log likelihood instead [log1-> 0 and as you go lower the log goes to -ve and then -ve infin]
```

```python
log_likelihood=0.0
n=0
# set n to normalize the nll 
for w in words[:3]:
    chs=['.'] + list(w) + ['.']
    for ch1, ch2 in zip(chs, chs[1:]):
        ix1=stoi[ch1]
        ix2=stoi[ch2]
        # probability that the models assign to each of this bigram
        prob=P[ix1, ix2]
        # log likelihood is the sum of the individual log probabilities
        logprob=torch.log(prob)
        log_likelihood+=logprob
        n+=1
        print(f'{ch1}{ch2}: {prob:.4f} {logprob:.4f}')
print(f'{log_likelihood=}')
# law in loss tterms measn better, so we have to multiply everything by -ve
nll=-log_likelihood
print(f"{nll=}")
print(f"Average of the nll: {nll/n}")
```
