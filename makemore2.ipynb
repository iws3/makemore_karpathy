```python
words=open("names.txt", "r").read().splitlines()
```

```python
#how often our characters preve-next appear in teh trainig corpox:

# BUilding a a bI GRAM
# WORKING WITH 2 CHARACTERS AT A TIME
b={}
for w in words:
    chs=['<S>'] + list(w) + ['<E>']
    for ch1, ch2 in zip(chs, chs[1:]):
        bigram=(ch1, ch2)
        # print(bigram)
        b[bigram]=b.get(bigram, 0) + 1
        # print(ch1, ch2)
b.items()       
```

# build a lookup table 

```python
joinEverything=''.join(words)
removeDuplicates=set(joinEverything)
charList=list(removeDuplicates)
sortedList=sorted(charList)
sortedList
```

```python
# GET RID OF THE SPECIAL CHARACTERS <s>and <e> with "." to solve te 
# spce issue
```

```python
# set <S>=26 and <E>=27
# instead <.>
```

```python
import torch

N= torch.zeros((27, 27), dtype=torch.int32)
N
```

```python
# offset index to start from 1 instead
# build a string to lsit dictionary
stoi={s:i+1 for i, s in enumerate(sortedList)}
stoi['.']=0
stoi
```

```python
# i to s remains the same:
# integer to string (itos)
itos={i:s for s, i in stoi.items()}
itos
```

```python
# better wasy to visualize our table using matplotlib
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.figure(figsize=(16, 16))
plt.imshow(N, cmap='Blues')

for i in range(28):
    for j in range(28):
        chstr=itos[i] + itos[j]
        plt.text(j, i, chstr, ha='center', va='bottom', color='gray')
        plt.text(j, i, N[i, j].item(), ha='center', va='top', color='gray')
plt.axis('on')
```
